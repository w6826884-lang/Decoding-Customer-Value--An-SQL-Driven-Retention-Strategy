### RFM Score

```python
data_1['R_Score'] = pd.qcut(data_1['Avg_no_of_yearly_purchases'], 5, labels=[1,2,3,4,5])
data_1['R_Score'] = data_1['R_Score'].astype(str)
```

```python
data_1['F_Score']=pd.qcut(data_1['Previous Purchases'], 5, labels=[1,2,3,4,5])
data_1['F_Score'] = data_1['F_Score'].astype(str)
```

```python
data_1['M_Score'] = pd.qcut(data_1['Purchase Amount (USD)'],5 ,labels=[1,2,3,4,5])
data_1['M_Score'] = data_1['M_Score'].astype(str)
```

```python
data_1['RFM_Score']= data_1['R_Score']+data_1['F_Score']+data_1['M_Score']
```

```python
data_1['R_Score'] = data_1['R_Score'].astype(int)
```

```python
data_1['F_Score'] = data_1['F_Score'].astype(int)
```

```python
data_1['M_Score'] = data_1['M_Score'].astype(int)
```

```python
def _assign_segment(row):
    r = row['R_Score']
    f = row['F_Score']
    m = row['M_Score']
    avg = (r + f + m) / 3

    if r >= 4 and f >= 4 and m >= 4:
        return 'Champion'
    if f >= 4 and r >= 3:
        return 'Loyal Customer'
    if r >= 4 and f >= 2 and f < 4:
        return 'Potential Loyalist'
    if r >= 4 and f <= 2:
        return 'Recent Customer'
    if r == 3 and f >= 2:
        return 'Promising'
    if r == 3 and f == 1 and m >= 3:
        return 'Need Attention'
    if r == 2 and f <= 2:
        return 'About to Sleep'
    if r <= 2 and f >= 3 and m >= 3:
        return 'At Risk'
    if r == 1 and f >= 4 and m >= 4:
        return "Can't Lose Them"
    if r <= 2 and f <= 2 and m <= 2:
        return 'Hibernating'
    if avg < 1.8:
        return 'Lost'

    return 'Need Attention'
```

```python
data_1['Customer_Segment']=data_1.apply(_assign_segment, axis = 1)
```

```python
data_1[['Purchase Amount (USD)','Previous Purchases','Avg_no_of_yearly_purchases']]= scalar.inverse_transform(data_1[['Purchase Amount (USD)','Previous Purchases','Avg_no_of_yearly_purchases']])
```

```python
import pandas as pd

# Load the dataframe
df = pd.read_csv('engineered_features_final.csv')

# 1. Products Table
products = df[['Item Purchased', 'Category']].drop_duplicates().reset_index(drop=True)
products['Item_ID'] = products.index + 1
# Reorder columns
products = products[['Item_ID', 'Item Purchased', 'Category']]

# 2. Customers Table
customer_cols = ['Customer ID', 'Age', 'Location', 'Subscription Status', 'Previous Purchases', 
                 'Frequency of Purchases', 'Avg_no_of_yearly_purchases', 'Aggregate', 
                 'Depandancy_Score', 'aggval', 'Value', 'Satisfaction_flag', 'RFM_Score', 'Customer_Segment']
customers = df[customer_cols].drop_duplicates()

# 3. Transactions Table
# Merge to get Item_ID back into the main dataframe
transactions_temp = df.merge(products, on=['Item Purchased', 'Category'])
transactions_temp['Transaction_ID'] = transactions_temp.index + 1

transaction_cols = ['Transaction_ID', 'Customer ID', 'Item_ID', 'Purchase Amount (USD)', 'Season', 
                    'Review Rating', 'Shipping Type', 'Discount Applied', 'Promo Code Used']
transactions = transactions_temp[transaction_cols]

# Save to CSV
products.to_csv('products_2nf.csv', index=False)
customers.to_csv('customers_2nf.csv', index=False)
transactions.to_csv('transactions_2nf.csv', index=False)

print("Files created successfully.")
```
